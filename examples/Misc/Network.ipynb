```python
def list_to_string(str_list, delimiter=","):
    return delimiter.join(map(str, str_list))


def two_dimensional_list_to_string(two_dim_list):
    str_list = []
    for row in two_dim_list:
        str_list.append(list_to_string(row))
    return list_to_string(str_list, "\n")
```

SGraph does not support undirected graphs out of the box.  
To create directed graphs use `combinations` and for undirected `permutations`.

```python
edge_list = []
ten_per = len(papers) //10
# os.remove(graph_path)
with open(graph_path, mode="a") as f:
    for i, line in enumerate(tqdm(papers["Authors List Sorted"])):
        if line and len(line) >1:
            edge_list += list(permutations(line,2))
        if i % ten_per == 0:
            f.write(two_dimensional_list_to_string([(int(edge[0]),int(edge[1])) for edge in edge_list])+"\n")
            edge_list = []

    f.write(two_dimensional_list_to_string(edge_list))
    edge_list = []
```

```python
author_names = mag.author_names
```

```python
vertices = g.vertices.join(author_names, on={"__id": "AuthorId"})
```

```python
# author_id_attr = vertices.pack_columns(["NormalizedName"], dtype=dict, new_column_name="attr")[["__id","attr"]].unstack(["__id","attr"])["Dict of __id_attr"][0]
```

```python
author_id_attr = {line["__id"]: line["NormalizedName"] for line in tqdm(vertices)}
```

```python
author_id_attr = defaultdict(lambda: None, author_id_attr)
```

```python
g.vertices["Author name"] = g.vertices["__id"].apply(lambda x: author_id_attr[x])
```

# Ciation Graph

```python
papers["First Author"] = papers["Authors List Sorted"].apply(lambda x: x[0])
```

```python
mag.references
```

```python
citations = mag.references.join(papers[["PaperId","First Author"]])
```

```python
citations = citations.join(papers[["PaperId","First Author"]],on={"PaperReferenceId":"PaperId"})
```

```python
citations = citations.rename({"First Author":"__src_id","First Author.1":"__dst_id"})
```

```python
citations["__src_id"] = citations["__src_id"].astype(int)
```

```python
citations["__dst_id"] = citations["__dst_id"].astype(int)
```

```python
citations_g = SGraph(edges=citations.dropna())
```

```python
citations_g.vertices["Author name"] = citations_g.vertices["__id"].apply(lambda x: author_id_attr[x])
```

```python
citations_g.save("citations.sgraph")
```
