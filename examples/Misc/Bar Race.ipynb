```python
def convert_to_barchart_format(promed, year_sf, count_column="count"):
    res_sf = SFrame()
    for d in promed["disease_name"].unique():
        temp_sf = SFrame()
        temp_sf["disease_name"] = [d]*len(year_sf)
        temp_sf["year"] = year_sf["year"]
        res_sf = res_sf.append(temp_sf)
    promed = promed.join(res_sf,how="right").sort("year")
    promed = promed.fillna(count_column,0)
    df = promed.to_dataframe()
    df = df.sort_values(["disease_name","year"])
    df['value'] = df.groupby(["disease_name"])[count_column].cumsum()
    df["lastValue"] = df.groupby(["disease_name"])["value"].shift(1)
    df = df.fillna(0)
    df["rank"] =df.groupby(["year"])["value"].rank(ascending=False)
    df[df["year"]==2017].sort_values('rank')
    return df.rename(columns={"disease_name":"name"})
```
